```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# CSV-Datei einlesen
df = pd.read_csv(
    r"source/achsen.csv",
    sep="\t",
    header=None,
    names=["Time", "X", "Y", "Z"]
)

# Zeit von Millisekunden in Sekunden umrechnen
df["Time"] = (df["Time"] - df["Time"].iloc[0]) / 1000

# Zeitdifferenzen zwischen den Messpunkten berechnen
dt = df["Time"].diff().dropna()

# Abtastfrequenz für jeden Schritt berechnen
fs = 1 / dt

# Mittelwert und Standardabweichung ausgeben
print("Abtastfrequenz Mittelwert:", fs.mean(), "Hz")
print("Abtastfrequenz Standardabweichung:", fs.std(), "Hz")

# Abtastfrequenz darstellen
plt.figure(figsize=(12, 6))

plt.plot(df["Time"].iloc[1:], fs, color="blue", label="Abtastfrequenz")
plt.axhline(fs.mean(), color="red", linestyle="--", label="Mittelwert")

plt.title("Abtastfrequenz (Aufgabe 8)")
plt.xlabel("Time in s")
plt.ylabel("Abtastfrequenz in Hz")

plt.legend()
plt.grid()
plt.tight_layout()
#plt.savefig("Aufgabe_8_Abbildung.png", dpi=300)
plt.show()
```

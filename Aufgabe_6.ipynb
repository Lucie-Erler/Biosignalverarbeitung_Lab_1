```python

import pandas as pd
import matplotlib.pyplot as plt

# CSV-Datei einlesen
df = pd.read_csv(
    r"C:\Users\lucie\Documents\Biosignalverarbeitung\Lab1\source\achsen.csv",
    sep="\t",
    header=None,
    names=['Time', 'X', 'Y', 'Z']
)

# Zeit von Millisekunden in Sekunden umrechnen
df['Time'] = (df['Time'] - df['Time'].iloc[0]) / 1000

# Beschleunigung in g umrechnen
df['X'] = df['X'] / 1024
df['Y'] = df['Y'] / 1024
df['Z'] = df['Z'] / 1024

# Abtastfrequenz berechnen
dt = df['Time'].diff().dropna()
fs = 1 / dt.mean()

print("Abtastfrequenz:", fs, "Hz")
print("Standardabweichung:", dt.std(), "s")

# Graph darstellen
plt.figure(figsize=(10, 6))

plt.plot(df['Time'], df['X'], color='red', label='X')
plt.plot(df['Time'], df['Y'], color='green', label='Y')
plt.plot(df['Time'], df['Z'], color='blue', label='Z')

plt.title("Accelerometer Test - 6 Orientierungen")
plt.xlabel("Time in s")
plt.ylabel("Acceleration in g")

plt.xlim(0, 60)
plt.ylim(-2, 2)

plt.legend()
plt.grid()
plt.tight_layout()
plt.show()
```
